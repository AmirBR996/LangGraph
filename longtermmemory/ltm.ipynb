```python
from langgraph.store.memory import InMemoryStore
```

```python
store = InMemoryStore()
```

```python
namespace = ("user", "u1")
```

### Creating Memories

```python
store.put(namespace, "1", {"data": "User like momo"})
store.put(namespace, "2", {"data": "User prefers warm mode"})
```

```python
namespace2 = ("user", "u2")
store.put(namespace2, "1", {"data": "User like chowmein"})
store.put(namespace2, "2", {"data": "User prefers android over ios"})
```

### Retrieving Memories

```python
store.get(namespace2, "2")
```

### Retrieving all Memories

```python
items = store.search(namespace2)

for item in items:
    print(item.value)
```

### Semantic Search

```python
from langchain_community.embeddings import HuggingFaceEmbeddings
from dotenv import load_dotenv

load_dotenv()

embeddings = HuggingFaceEmbeddings()
```

```python
store = InMemoryStore(index={'embed': embeddings, 'dims':1536})
```

```python
namespace = ('users', 'u1')
```

```python
store.put(namespace, "1", {"data": "User is based in Nepal"})
store.put(namespace, "2", {"data": "User lives in Kathmandu, Nepal"})
store.put(namespace, "3", {"data": "User is interested in AI and machine learning"})
store.put(namespace, "4", {"data": "User is building AI voice agents"})
store.put(namespace, "5", {"data": "User is learning LangGraph"})
store.put(namespace, "6", {"data": "User is exploring PostgreSQL for LangGraph memory"})
store.put(namespace, "7", {"data": "User is working on FastAPI-based AI applications"})
store.put(namespace, "8", {"data": "User is interested in deploying AI applications"})
store.put(namespace, "9", {"data": "User wants to build production-ready AI systems"})
store.put(namespace, "10", {"data": "User is preparing for an AI/ML engineering career in Nepal"})
```

```python
items = store.search(namespace, query="what is the user currently learning", limit=1)

for item in items:
    print(item.value)
```

```python
items = store.search(namespace, query="what are user's preferences", limit=3)

for item in items:
    print(item.value)
```
